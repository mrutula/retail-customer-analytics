```python
codes = ["M", 'D', "C2", "DOT", "POST", "PADS", "BANK CHARGES"]

restored_df[restored_df["StockCode"].isin(codes)].groupby(
    ["StockCode", "Description"]
).agg(
    rows=("StockCode", "size"),
    total_quantity=("Quantity", "sum"),
    min_price=("Price", "min"),
    max_price=("Price", "max"),
    customers=("Customer ID", "nunique")
).sort_index()
```

Cleaning decision
1) BANK CHARGES - Almost always standalone and doesnt look like a transaction - REMOVE
2) CARRIAGE - Mostly with other invoices - not a transation - REMOVE
3) Discount - only 5 records with standalone invoices - not a transaction - REMOVE
4) DOTCOM POSTAGE - Always with other transation - not a transaction - REMOVE 
5) Manual - Almost alwyas with other transaction - not a transation - REMOVE
6) PADS - padding related always with other invoice - not a transtion - REMOVE
7) POST - Posting related - not a transtion - REMOVE
